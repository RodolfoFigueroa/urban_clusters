```python
import os
from pathlib import Path

import geopandas as gpd
```

```python
population_grids_path = Path(os.environ["POPULATION_GRIDS_PATH"])

points_path = Path("./points")
points_path.mkdir(parents=True, exist_ok=True)
```

```python
df_mun = gpd.read_file(
    population_grids_path / "final" / "framework" / "mun" / "2020.gpkg",
)
```

```python
groups = [
    ["19039", "19019", "19048", "19018", "19021", "19046", "19026", "19006", "19031"],
    ["19009", "19041"],
    ["19004", "19049"],
    ["19045", "19012", "19025"],
    [
        "19039",
        "19019",
        "19048",
        "19018",
        "19021",
        "19046",
        "19026",
        "19006",
        "19031",
        "19009",
        "19041",
        "19004",
        "19049",
        "19045",
        "19012",
        "19025",
    ],
]

out = []
for i, group in enumerate(groups):
    temp = df_mun[df_mun["CVEGEO"].isin(group)].assign(
        geometry=lambda df: df["geometry"].buffer(10),
    )
    out = gpd.GeoDataFrame(geometry=[temp["geometry"].union_all()], crs=temp.crs)
    out.to_file(points_path / f"group_{i + 1}.gpkg")
```
